```python
def image_to_base64(image_path):
    """
    Loads an image from the given file path and converts it to a Base64 string.

    Args:
        image_path (str): The path to the image file.

    Returns:
        str: The Base64-encoded string of the image data.
    """
    with open(image_path, "rb") as image_file:
        # Read the image file in binary mode.
        image_data = image_file.read()
        # Encode the binary data to base64.
        base64_encoded = base64.b64encode(image_data).decode("utf-8")
    return base64_encoded
```

```python
def base64_to_image(base64_string, output_file_path):
    """
    Converts a Base64-encoded string to an image and saves it to disk.

    Args:
        base64_string (str): The Base64 string representing the image.
        output_file_path (str): The path (including filename and extension) where the image will be saved.
    """
    # Decode the Base64 string to binary data.
    image_data = base64.b64decode(base64_string)
    # Write the binary data to a file.
    with open(output_file_path, "wb") as image_file:
        image_file.write(image_data)
    print(f"Image has been saved to {output_file_path}")
```

```python
import os
import random
import requests

# Base path for cloth types
PATH = r'C:\Project Design\testing\CatVTONTest\damitxx'

def get_cloth_type(cloth_type):
    cloth_path = os.path.join(PATH, cloth_type)
    # Build full paths for all files in the specified cloth type directory.
    cloth_list = [os.path.join(cloth_path, filename) for filename in os.listdir(cloth_path)]
    
    # Build full paths for all files in the people directory.
    people_list = [
        os.path.join(r'C:\Project Design\testing\CatVTONTest\people', filename)
        for filename in os.listdir(r'C:\Project Design\testing\CatVTONTest\people')
    ]
    
    return cloth_list, people_list

# Retrieve the lists for the cloth type 'Pants'
cloth_target = 'Shorts'
hehez = 'lower'
cloth_list, people_list = get_cloth_type(cloth_target)

# Check that it's possible to generate 20 unique pairs
max_possible_pairs = len(cloth_list) * len(people_list)
if max_possible_pairs < 20:
    raise ValueError(f"Not enough unique pairs possible. Maximum possible is {max_possible_pairs}.")

# Using a set to store unique (cloth, person) pairs.
unique_pairs = set()

# Loop until we have 20 unique pairs.
while len(unique_pairs) < 20:
    cloth_item = random.choice(cloth_list)
    people_item = random.choice(people_list)
    # The pair is added to the set (order matters since it's a tuple).
    unique_pairs.add((cloth_item, people_item))

# Convert the set into a list for further use (if needed)
unique_pairs = list(unique_pairs)

# Display the unique pairs
for i, pair in enumerate(unique_pairs, start=1):
    base64cloth = image_to_base64(pair[0])
    base64person = image_to_base64(pair[1])
    payload = {
        'cloth_image': base64cloth,
        'person_image': base64person,
        'cloth_type': hehez
    }
    
    response = requests.post('http://127.0.0.1:8888//tryon', json = payload)

    response = response.json()

    
    base64_to_image(response['result_image'], f'output//{cloth_target}//{i}.png')

    



```
